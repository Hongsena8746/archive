```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage
import os

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
    api_key=os.getenv("OPENAI_API_KEY"),  # ← 이거만 사용! (오타 금지)
)

messages = [
    SystemMessage(content="너는 '미녀와 야수'의 '벨' 역할이야. 캐릭터에 맞게 대화해."),
    HumanMessage(content="안녕하세요! 저는 게스트입니다. 오늘 저녁에 뭐 먹을까요?")
]

res = model.invoke(messages)
print(res.content)    
```
